````python
#Implement Matrix Multiplication using Map-Reduce

def mapper(matrix_A, matrix_B, m, n, p):
    mapped_results = []

    # Emit key-value pairs for matrix A
    for i in range(m):
        for j in range(n):
            for k in range(p):
                mapped_results.append(((i, k), ("A", j, matrix_A[i][j])))  # Fixed missing parenthesis

    # Emit key-value pairs for matrix B
    for j in range(n):
        for k in range(p):
            for i in range(m):
                mapped_results.append(((i, k), ("B", j, matrix_B[j][k])))  # Fixed missing parenthesis

    return mapped_results
````

````python
def shuffle_and_sort(mapped_results):
    shuffled_data = {}

    # Group values by key (i, k)
    for key, value in mapped_results:
        if key in shuffled_data:
            shuffled_data[key].append(value)
        else:
            shuffled_data[key] = [value]

    return shuffled_data
````

````python
def reducer(shuffled_data, n):
    result_matrix = {}

    # Compute the dot product for each key (i, k)
    for key, values in shuffled_data.items():
        i, k = key
        A = {}
        B = {}

        # Populate A and B dictionaries
        for value in values:
            matrix, j, val = value
            if matrix == "A":
                A[j] = val
            elif matrix == "B":
                B[j] = val

        # Compute the dot product
        result = 0
        for j in range(n):
            if j in A and j in B:
                result += A[j] * B[j]

        # Store the result in the result matrix
        result_matrix[(i, k)] = result

    return result_matrix
````

````python
# Matrix Multiplication using Map-Reduce

def mapper(matrix_A, matrix_B, m, n, p):
    mapped_results = []

    # Mapper for Matrix A
    for i in range(m):
        for j in range(n):
            for k in range(p):
                mapped_results.append(
                    ((i, k), ("A", j, matrix_A[i][j]))
                )

    # Mapper for Matrix B
    for j in range(n):
        for k in range(p):
            for i in range(m):
                mapped_results.append(
                    ((i, k), ("B", j, matrix_B[j][k]))
                )

    return mapped_results


# Shuffle and Sort Phase
def shuffle_and_sort(mapped_results):
    shuffled_data = {}

    for key, value in mapped_results:
        if key not in shuffled_data:
            shuffled_data[key] = []

        shuffled_data[key].append(value)

    return shuffled_data


# Reducer Function
def reducer(shuffled_data, n):
    result_matrix = {}

    for key, values in shuffled_data.items():

        A = {}
        B = {}

        # Separate A and B values
        for value in values:
            matrix_name, j, val = value

            if matrix_name == "A":
                A[j] = val

            elif matrix_name == "B":
                B[j] = val

        # Dot Product Calculation
        result = 0

        for j in range(n):
            if j in A and j in B:
                result += A[j] * B[j]

        result_matrix[key] = result

    return result_matrix


# Main Function
if __name__ == "__main__":

    matrix_A = [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]
    ]
    
    matrix_B = [
        [9, 8, 7],
        [6, 5, 4],
        [3, 2, 1]
    ]

    # Dimensions
    m = len(matrix_A)
    n = len(matrix_A[0])
    p = len(matrix_B[0])

    # Map Phase
    mapped_results = mapper(matrix_A, matrix_B, m, n, p)

    # Shuffle and Sort Phase
    shuffled_data = shuffle_and_sort(mapped_results)

    # Reduce Phase
    result_matrix = reducer(shuffled_data, n)

    # Print Result Matrix
    print("Resultant Matrix C:")

    for i in range(m):
        for k in range(p):
            print(f"C[{i}][{k}] = {result_matrix[(i, k)]}")
````

````python
# # Matrix Multiplication using Map-Reduce

# ## Definition

# Matrix multiplication is the process of multiplying two matrices to produce a new matrix.

# If:

# [
# A_{m \times n} \times B_{n \times p} = C_{m \times p}
# ]

# then each element is calculated as:

# [
# C[i][k] = \sum_{j=0}^{n-1} A[i][j] \times B[j][k]
# ]

# ---

# # What is Map-Reduce?

# Map-Reduce is a programming model used for processing large amounts of data in parallel and distributed systems. It mainly consists of:

# 1. Map Phase
# 2. Shuffle and Sort Phase
# 3. Reduce Phase

# It is widely used in Apache Hadoop and Big Data applications.

# ---

# # Matrix Multiplication using Map-Reduce

# ## 1. Map Phase

# The mapper generates key-value pairs for matrix elements.

# Key:
# [
# (i,k)
# ]

# Values:

# * From Matrix A:
#   [
#   ("A", j, A[i][j])
#   ]

# * From Matrix B:
#   [
#   ("B", j, B[j][k])
#   ]

# The mapper sends all values needed to compute one element of the result matrix.

# ---

# ## 2. Shuffle and Sort Phase

# In this phase, all values having the same key ((i,k)) are grouped together.

# Example:

# For key ((0,0)):

# ```text id="5l5lyu"
# ('A',0,1)
# ('A',1,2)
# ('B',0,5)
# ('B',1,7)
# ```

# This grouping helps the reducer perform multiplication correctly.

# ---

# ## 3. Reduce Phase

# The reducer matches elements with the same index (j), multiplies them, and calculates their sum.

# Example:

# [
# C[0][0] = (1 \times 5) + (2 \times 7)
# ]

# [
# = 5 + 14 = 19
# ]

# The final value is stored in the result matrix.

# ---

# # Flow Diagram

# ```text id="0slk9u"
# Matrix A + Matrix B
#           |
#         Mapper
#           |
#    Key-Value Pairs
#           |
#    Shuffle & Sort
#           |
#         Reducer
#           |
#      Result Matrix
# ```

# ---

# # 3×3 Matrix Example

# [
# A =
# \begin{bmatrix}
# 1 & 2 & 3 \
# 4 & 5 & 6 \
# 7 & 8 & 9
# \end{bmatrix}
# ]

# [
# B =
# \begin{bmatrix}
# 9 & 8 & 7 \
# 6 & 5 & 4 \
# 3 & 2 & 1
# \end{bmatrix}
# ]

# Result:

# [
# C =
# \begin{bmatrix}
# 30 & 24 & 18 \
# 84 & 69 & 54 \
# 138 & 114 & 90
# \end{bmatrix}
# ]

# ---

# # Advantages

# * Supports parallel processing
# * Faster computation for large matrices
# * Scalable for Big Data systems
# * Efficient in distributed environments

# ---

# # Disadvantages

# * High communication overhead
# * Complex implementation
# * Not efficient for small datasets

# ---

# # Applications

# * Big Data Analytics
# * Scientific Computing
# * Machine Learning
# * Image Processing

# ---

# # Conclusion

# Map-Reduce performs matrix multiplication by dividing computation into Map, Shuffle,
# and Reduce phases. The mapper generates key-value pairs, shuffle groups similar 
# keys, and reducer computes the final matrix values. This method is efficient for large-scale distributed computing systems like Hadoop.
````
