```python
A = dict()
B = dict()
A_union_B = dict()

A = {'a': 0.2, 'b': 0.7, 'c': 0.6, 'd': 0.6}
B = {'a': 0.4, 'b': 0.3, 'c': 0.8, 'd': 0.6}

print("Fuzzy Set A is: ", A)
print("Fuzzy Set B is: ", B)

for A_key,B_key in zip(A,B):
    A_value = A[A_key]
    B_value = B[B_key]

    A_union_B[A_key] = max(A_value, B_value)

print("Union of A and B is: ", A_union_B)
```

```python
A = dict()
B = dict()
A_intersect_B = dict()

A = {'a': 0.2, 'b': 0.7, 'c': 0.6, 'd': 0.6}
B = {'a': 0.4, 'b': 0.3, 'c': 0.8, 'd': 0.6}

print("Fuzzy Set A is: ", A)
print("Fuzzy Set B is: ", B)

for A_key,B_key in zip(A,B):
    A_value = A[A_key]
    B_value = B[B_key]

    A_intersect_B[A_key] = min(A_value, B_value)

print("Intersect of A and B is: ", A_intersect_B)
```

```python
A = dict()
B = dict()
A_complement = dict()

A = {'a': 0.2, 'b': 0.7, 'c': 0.6, 'd': 0.6}

print("Fuzzy Set A is: ", A)

for A_key in A:
    A_value = A[A_key]

    A_complement[A_key] = 1 - A_value

print("Complement of A is: ", A_complement)
```

```python
A = dict()
B = dict()
A_difference_B = dict()

A = {'a': 0.2, 'b': 0.3, 'c': 0.6, 'd': 0.6}
B = {'a': 0.9, 'b': 0.9, 'c': 0.8, 'd': 0.6}

print("Fuzzy Set A is: ", A)
print("Fuzzy Set B is: ", B)

for A_key,B_key in zip(A,B):
    A_value = A[A_key]
    B_value = B[B_key]
    #B_value = 1 - B_value

    A_difference_B[A_key] = min(A_value,(1-B_value))

print("Difference of A and B is: ", A_difference_B)
```

```python
A = dict()
B = dict()
A_cartesian_B = dict()

A = {'a': 0.2, 'b': 0.3}
B = {'c': 0.8, 'd': 0.6}

for A_key in A:
    A_value = A[A_key]
    for B_key in B:
        B_value = B[B_key]
        A_cartesian_B[(A_key,B_key)] = min(A_value,B_value)

print("Cartesian Product of A and B is: ", A_cartesian_B)
```

```python
def max_min_composition(R, S):
    result = {}
    for (x, y1) in R:
        for (y2, z) in S:
            if y1 == y2:
                key = (x, z)
                value = min(R[(x, y1)], S[(y2, z)])
                if key in result:
                    result[key] = max(result[key], value)
                else:
                    result[key] = value
    return result

#Fuzzy relations
R = {('x1', 'y1'): 0.3, ('x1', 'y2'): 0.7}
S = {('y1', 'z1'): 0.5, ('y2', 'z1'): 0.8}
print("\nMax-Min Composition (R ∘ S):", max_min_composition(R, S))
```

```python
# 1) Theory of Fuzzy Sets

# A fuzzy set is a set in which an element can belong partially, not just fully or not at all.

# In a normal set:

# membership is either 0 or 1

# In a fuzzy set:

# membership can be any value between 0 and 1

# Example:

# A = {a: 0.2, b: 0.7, c: 0.6}

# This means:

# a belongs to set A with degree 0.2
# b belongs with degree 0.7
# c belongs with degree 0.6
# 2) Important Definitions
# a) Union of fuzzy sets

# The union gives the maximum membership.

# Formula:

# μ(A ∪ B)(x) = max(μA(x), μB(x))
# Example

# If:

# A(a) = 0.2
# B(a) = 0.4

# Then:

# Union(a) = max(0.2, 0.4) = 0.4
# b) Intersection of fuzzy sets

# The intersection gives the minimum membership.

# Formula:

# μ(A ∩ B)(x) = min(μA(x), μB(x))
# Example

# If:

# A(b) = 0.7
# B(b) = 0.3

# Then:

# Intersection(b) = min(0.7, 0.3) = 0.3
# c) Complement of fuzzy set

# The complement means the degree of non-membership.

# Formula:

# μ(A')(x) = 1 - μA(x)
# Example

# If:

# A(c) = 0.6

# Then:

# Complement(c) = 1 - 0.6 = 0.4
# d) Difference of fuzzy sets

# Difference is usually defined as:

# A - B = A ∩ B'

# So the formula becomes:

# μ(A - B)(x) = min(μA(x), 1 - μB(x))
# Example

# If:

# A(d) = 0.6
# B(d) = 0.6

# Then:

# Difference(d) = min(0.6, 1 - 0.6) = min(0.6, 0.4) = 0.4
# 3) Fuzzy Relation and Cartesian Product

# A fuzzy relation between two fuzzy sets A and B is formed on the Cartesian product A × B.

# For each pair (x, y):

# μ(A × B)(x, y) = min(μA(x), μB(y))

# This gives the membership of the pair in the relation.

# Example

# If:

# A(a) = 0.2
# B(c) = 0.8

# Then:

# (a, c) = min(0.2, 0.8) = 0.2
# 4) Max-Min Composition of Fuzzy Relations

# If:

# R is a relation from X to Y
# S is a relation from Y to Z

# Then the composition R ∘ S is defined by:

# μ(R ∘ S)(x, z) = max over y [ min( μR(x, y), μS(y, z) ) ]
# Meaning

# For each (x, z):

# take all possible intermediate y
# compute the min
# then take the max of those values

# This is called max-min composition.

# 5) Code Explanation and Correctness
# A) Union code
# A = {'a': 0.2, 'b': 0.7, 'c': 0.6, 'd': 0.6}
# B = {'a': 0.4, 'b': 0.3, 'c': 0.8, 'd': 0.6}

# for A_key,B_key in zip(A,B):
#     A_value = A[A_key]
#     B_value = B[B_key]
#     A_union_B[A_key] = max(A_value, B_value)
# Explanation
# zip(A, B) pairs keys from A and B in the same order.
# For each key, it takes the max of the two membership values.
# Correctness

# ✅ The logic is correct for fuzzy union.

# Small improvement

# Using zip(A, B) works here only because both dictionaries have the same keys in the same insertion order. A safer version is:

# for key in A:
#     A_union_B[key] = max(A[key], B[key])
# B) Intersection code
# for A_key,B_key in zip(A,B):
#     A_value = A[A_key]
#     B_value = B[B_key]
#     A_intersect_B[A_key] = min(A_value, B_value)
# Explanation
# Takes minimum of membership values for each key.
# Correctness

# ✅ Correct for fuzzy intersection.

# Improvement

# Same as above: better to use one key variable.

# C) Complement code
# for A_key in A:
#     A_value = A[A_key]
#     A_complement[A_key] = 1 - A_value
# Explanation
# For each element, subtracts membership from 1.
# Correctness

# ✅ Correct.

# Example

# If A = {a: 0.2}, then complement of a is 0.8.

# D) Difference code
# for A_key,B_key in zip(A,B):
#     A_value = A[A_key]
#     B_value = B[B_key]
#     A_difference_B[A_key] = min(A_value,(1-B_value))
# Explanation
# Computes fuzzy difference as A ∩ B'
# So it uses min(A_value, 1 - B_value)
# Correctness

# ✅ Correct.

# Note

# This is not ordinary subtraction. In fuzzy logic, difference is usually defined this way.

# E) Cartesian product code
# A = {'a': 0.2, 'b': 0.3}
# B = {'c': 0.8, 'd': 0.6}

# for A_key in A:
#     A_value = A[A_key]
#     for B_key in B:
#         B_value = B[B_key]
#         A_cartesian_B[(A_key,B_key)] = min(A_value,B_value)
# Explanation
# Generates all pairs from A and B
# Membership of each pair is the minimum of the two memberships
# Correctness

# ✅ Correct for fuzzy Cartesian product / fuzzy relation.

# Example result
# (a, c) = min(0.2, 0.8) = 0.2
# (b, d) = min(0.3, 0.6) = 0.3
# F) Max-min composition code
# def max_min_composition(R, S):
#     result = {}
#     for (x, y1) in R:
#         for (y2, z) in S:
#             if y1 == y2:
#                 key = (x, z)
#                 value = min(R[(x, y1)], S[(y2, z)])
#                 if key in result:
#                     result[key] = max(result[key], value)
#                 else:
#                     result[key] = value
#     return result
# Explanation
# It checks where the middle element matches: y1 == y2
# Then computes min(R, S)
# If there are multiple paths for the same (x, z), it keeps the max
# Correctness

# ✅ Correct implementation of max-min composition.
```
