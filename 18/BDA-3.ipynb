```python
import csv

# Read data from CSV file
with open('grade.csv', 'r') as file:
    reader = csv.DictReader(file)
    original_data = list(reader)

# Function to assign grade based on marks
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    elif marks >= 60:
        return 'D'
    else:
        return 'F'

# Mapper Phase
def mapper(data):
    mapped_data = []
    for row in data:
        student_id = row['StudentID']
        name = row['Name']
        marks = int(row['Marks'])
        mapped_data.append((student_id, name, marks))
    return mapped_data

# Reducer Phase
def reducer(mapped_data):
    reduced_data = []
    for student_id, name, marks in mapped_data:
        grade = assign_grade(marks)
        reduced_data.append((student_id, name, marks, grade))
    return reduced_data

# Execute MapReduce steps
mapped = mapper(original_data)
reduced = reducer(mapped)

# Display original dataset
print("Original Dataset:")
print("StudentID\tName\t\tMarks")
for row in original_data:
    print(f"{row['StudentID']}\t\t{row['Name']}\t\t{row['Marks']}")

# Display final result
print("\nDataset with Grades:")
print("StudentID\tName\t\tMarks\tGrade")
for student_id, name, marks, grade in reduced:
    print(f"{student_id}\t\t{name}\t\t{marks}\t{grade}")
```

```python
# Theory: MapReduce to Find Grades of Students
# Definition

# MapReduce is a programming model used to process large datasets in parallel. It works in two main phases: Map and Reduce. It is commonly used in distributed systems and big data processing.

# Working of MapReduce
# 1. Mapper

# The mapper reads each student record from the input file and converts it into an intermediate key-value format.

# For example:

# (StudentID,(Name,Marks))

# In this problem, the mapper does not calculate the grade. It only prepares the data for processing.

# 2. Shuffle and Sort

# This phase groups all records with the same key together. In a real distributed system, this is handled automatically by the MapReduce framework.

# For student grading, each student record is independent, so grouping is simple.

# 3. Reducer

# The reducer receives the mapped data and performs the final operation. Here, it checks the marks and assigns the grade:

# 90 and above → A
# 80 to 89 → B
# 70 to 79 → C
# 60 to 69 → D
# Below 60 → F

# Thus, the reducer produces the final output with grades.

# How MapReduce is Used Here

# In this problem, MapReduce is used as a record-wise transformation model.

# Map phase: reads student details and converts them into intermediate form.
# Reduce phase: assigns grades based on marks.

# This is not a heavy aggregation problem like word count, but it still follows the MapReduce structure because each record is processed independently and can be distributed across multiple nodes.

# Why MapReduce is Suitable
# It can process many student records in parallel.
# It is simple and scalable.
# It separates data reading and final computation.
# It is useful when the dataset is large.
```
