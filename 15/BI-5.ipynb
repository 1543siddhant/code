```python
#Perform the data classification algorithm using any Classification algorithm

# Importing necessary libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Loading the Iris dataset
iris = load_iris()
X = iris.data  # Features
y = iris.target  # Target labels

# Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Creating and training the Decision Tree classifier
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

# Making predictions
y_pred = clf.predict(X_test)

# Evaluating the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB

# Naive Bayes Classifier
nb = GaussianNB()
nb.fit(X_train, y_train)

# Predict
y_pred = nb.predict(X_test)

# Evaluate
print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
Data Classification using Classification Algorithms

Aim:
To perform data classification using machine learning classification algorithms and evaluate model performance.

Theory:

Classification is a supervised machine learning technique used to predict categorical labels or classes based on input data. The model learns patterns from labeled training data and predicts the class of new unseen data.

In this program, the Iris dataset is used. The dataset contains flower measurements such as:

* Sepal Length
* Sepal Width
* Petal Length
* Petal Width

The target classes are:

* Iris Setosa
* Iris Versicolor
* Iris Virginica

The program uses:

1. Decision Tree Classifier
2. Naive Bayes Classifier

Code Explanation

1. Import Libraries

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

These libraries are used for:

* Loading dataset
* Splitting data
* Training classifier
* Evaluating performance

2. Load Dataset

iris = load_iris()
X = iris.data
y = iris.target

* X contains input features.
* y contains target labels.

3. Split Dataset

X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.3, random_state=42
)

* 70% data is used for training.
* 30% data is used for testing.

4. Decision Tree Classifier

clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

* Creates Decision Tree model.
* fit() trains the model.

5. Prediction

y_pred = clf.predict(X_test)

Predicts output classes for test data.

6. Model Evaluation

accuracy_score(y_test, y_pred)

Measures prediction accuracy.

classification_report()

Displays:

* Precision
* Recall
* F1-score

confusion_matrix()

Shows correct and incorrect classifications.

7. Naive Bayes Classifier

from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(X_train, y_train)

Gaussian Naive Bayes classifier is trained using probability theory.

Explanation of Output

Naive Bayes Accuracy:
0.9777777777777777

This means:

* The model correctly classified approximately 97.78% of test samples.

Classification Report:

1. Precision
   Measures how many predicted values were actually correct.

Formula:

Precision = TP / (TP + FP)

2. Recall
   Measures how many actual values were correctly predicted.

Formula:

Recall = TP / (TP + FN)

3. F1-score
   Harmonic mean of precision and recall.

4. Support
   Number of actual samples in each class.

Confusion Matrix:

[[19 0 0]
[0 12 1]
[0 0 13]]

Explanation:

* 19 samples of class 0 classified correctly.
* 12 samples of class 1 classified correctly.
* 1 sample of class 1 misclassified as class 2.
* 13 samples of class 2 classified correctly.

This indicates very high model performance.

Benefits of Classification

* Automatic prediction
* Fast decision making
* High accuracy
* Useful for large datasets
* Helps in pattern recognition

Applications

* Email Spam Detection
* Disease Prediction
* Fraud Detection
* Customer Classification
* Image Recognition

Additional Classification Algorithms

1. Logistic Regression

Theory:
Logistic Regression is used for binary and multiclass classification. It predicts probabilities using sigmoid function.

Code Snippet:

from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

2. K-Nearest Neighbors (KNN)

Theory:
KNN classifies data based on nearest neighboring data points.

Code Snippet:

from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

3. Support Vector Machine (SVM)

Theory:
SVM separates classes using an optimal hyperplane.

Code Snippet:

from sklearn.svm import SVC

model = SVC()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

Clustering Algorithms

Clustering is an unsupervised machine learning technique used to group similar data points together without labeled output.

1. K-Means Clustering

Theory:
K-Means divides data into K clusters based on centroid distance.

Code Snippet:

from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

labels = kmeans.labels_

Applications:

* Customer segmentation
* Market analysis

2. Hierarchical Clustering

Theory:
Hierarchical clustering creates clusters in tree-like structure called dendrogram.

Code Snippet:

from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=3)

labels = hc.fit_predict(X)

Applications:

* Biological data analysis
* Document clustering

3. DBSCAN

Theory:
DBSCAN groups dense regions and identifies noise points.

Code Snippet:

from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.5, min_samples=5)

labels = dbscan.fit_predict(X)

Applications:

* Anomaly detection
* Spatial data analysis

Difference Between Classification and Clustering

Classification:

* Supervised learning
* Uses labeled data
* Predicts predefined classes

Clustering:

* Unsupervised learning
* Uses unlabeled data
* Groups similar data automatically

Conclusion

The program successfully performs data classification using Decision Tree and Naive Bayes algorithms on the Iris dataset. The model achieves high accuracy and correctly predicts flower classes. Additional classification and clustering algorithms provide different approaches for supervised and unsupervised learning tasks in machine learning.
```
